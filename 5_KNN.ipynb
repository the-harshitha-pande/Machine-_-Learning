```python
#Euclidian distance - without inbuilt functions
import pandas as pd
import math

# Step 1: Load Iris dataset (only sepal length, sepal width, species)
from sklearn.datasets import load_iris
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = [iris.target_names[i] for i in iris.target]

# Keep only sepal length, sepal width, species
df = df[['sepal length (cm)', 'sepal width (cm)', 'species']]
print("Initial Table:")
df = df.sample(frac=1, random_state=42).reset_index(drop=True)
print(df.head(10))   # now you’ll see a mix of species

# Step 2: Ask user for input values
user_sepal_length = float(input("Enter sepal length: "))
user_sepal_width  = float(input("Enter sepal width: "))

# Step 3: Calculate Euclidean distance manually
distances = []
for i in range(len(df)):
    sl = df.iloc[i,0]
    sw = df.iloc[i,1]
    d = math.sqrt((user_sepal_length - sl)**2 + (user_sepal_width - sw)**2)
    distances.append(d)

df['distance'] = distances

# Step 4: Rank based on distance
df['rank'] = df['distance'].rank(method='first')  # smaller distance = smaller rank
df_sorted = df.sort_values(by='distance').reset_index(drop=True)

print("\nTable with Distance and Rank:")
print(df_sorted[['sepal length (cm)','sepal width (cm)','species','distance','rank']].head(10))

# Step 5: Choose k and predict species
k = int(input("Enter k value: "))
top_k = df_sorted.head(k)

# Count species frequency among top k
species_counts = top_k['species'].value_counts()
predicted_species = species_counts.idxmax()

print("\nTop", k, "neighbors:")
print(top_k[['sepal length (cm)','sepal width (cm)','species','distance','rank']])

print("\nPredicted species for input (", user_sepal_length, ",", user_sepal_width, ") is:", predicted_species)
```

```python
#Manhattan distance:
import pandas as pd

# Step 1: Load Iris dataset (only sepal length, sepal width, species)
from sklearn.datasets import load_iris
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = [iris.target_names[i] for i in iris.target]

# Keep only sepal length, sepal width, species
df = df[['sepal length (cm)', 'sepal width (cm)', 'species']]

# Shuffle so species are mixed
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

print("Initial Table (mixed species):")
print(df.head(10))

# Step 2: Ask user for input values
user_sepal_length = float(input("Enter sepal length: "))
user_sepal_width  = float(input("Enter sepal width: "))

# Step 3: Calculate Manhattan distance manually
distances = []
for i in range(len(df)):
    sl = df.iloc[i,0]
    sw = df.iloc[i,1]
    d = abs(user_sepal_length - sl) + abs(user_sepal_width - sw)
    distances.append(d)

df['distance'] = distances

# Step 4: Rank based on distance
df['rank'] = df['distance'].rank(method='first')
df_sorted = df.sort_values(by='distance').reset_index(drop=True)

print("\nTable with Manhattan Distance and Rank:")
print(df_sorted[['sepal length (cm)','sepal width (cm)','species','distance','rank']].head(10))

# Step 5: Choose k and predict species
k = int(input("Enter k value: "))
top_k = df_sorted.head(k)

# Count species frequency among top k
species_counts = top_k['species'].value_counts()
predicted_species = species_counts.idxmax()

print("\nTop", k, "neighbors:")
print(top_k[['sepal length (cm)','sepal width (cm)','species','distance','rank']])

print("\nPredicted species for input (", user_sepal_length, ",", user_sepal_width, ") is:", predicted_species)
```

```python
#Minkowski distance
import pandas as pd

# Step 1: Load Iris dataset (only sepal length, sepal width, species)
from sklearn.datasets import load_iris
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = [iris.target_names[i] for i in iris.target]

# Keep only sepal length, sepal width, species
df = df[['sepal length (cm)', 'sepal width (cm)', 'species']]

# Shuffle so species are mixed
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

print("Initial Table (mixed species):")
print(df.head(10))

# Step 2: Ask user for input values
user_sepal_length = float(input("Enter sepal length: "))
user_sepal_width  = float(input("Enter sepal width: "))

# Step 3: Ask for Minkowski order p
p = int(input("Enter Minkowski order p (e.g., 1=Manhattan, 2=Euclidean): "))

# Step 4: Calculate Minkowski distance manually
distances = []
for i in range(len(df)):
    sl = df.iloc[i,0]
    sw = df.iloc[i,1]
    d = (abs(user_sepal_length - sl)**p + abs(user_sepal_width - sw)**p)**(1/p)
    distances.append(d)

df['distance'] = distances

# Step 5: Rank based on distance
df['rank'] = df['distance'].rank(method='first')
df_sorted = df.sort_values(by='distance').reset_index(drop=True)

print("\nTable with Minkowski Distance (p =", p, ") and Rank:")
print(df_sorted[['sepal length (cm)','sepal width (cm)','species','distance','rank']].head(10))

# Step 6: Choose k and predict species
k = int(input("Enter k value: "))
top_k = df_sorted.head(k)

# Count species frequency among top k
species_counts = top_k['species'].value_counts()
predicted_species = species_counts.idxmax()

print("\nTop", k, "neighbors:")
print(top_k[['sepal length (cm)','sepal width (cm)','species','distance','rank']])

print("\nPredicted species for input (", user_sepal_length, ",", user_sepal_width, ") is:", predicted_species)
```

```python
#different k values
import pandas as pd
import math
from sklearn.datasets import load_iris

# Step 1: Load Iris dataset (only sepal length, sepal width, species)
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = [iris.target_names[i] for i in iris.target]
df = df[['sepal length (cm)', 'sepal width (cm)', 'species']]

# Shuffle so species are mixed
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

print("Initial Table (mixed species):")
print(df.head(10))

# Step 2: Ask user for input values
user_sepal_length = float(input("Enter sepal length: "))
user_sepal_width  = float(input("Enter sepal width: "))

# Step 3: Calculate Euclidean distance manually
distances = []
for i in range(len(df)):
    sl = df.iloc[i,0]
    sw = df.iloc[i,1]
    d = math.sqrt((user_sepal_length - sl)**2 + (user_sepal_width - sw)**2)
    distances.append(d)

df['distance'] = distances

# Step 4: Rank based on distance
df['rank'] = df['distance'].rank(method='first')
df_sorted = df.sort_values(by='distance').reset_index(drop=True)

print("\nTable with Euclidean Distance and Rank:")
print(df_sorted[['sepal length (cm)','sepal width (cm)','species','distance','rank']].head(10))

# Step 5: Ask user for multiple k values
k_values = input("Enter k values separated by commas (e.g. 1,3,5): ")
k_values = [int(k.strip()) for k in k_values.split(",")]

# Step 6: Predictions for each k
for k in k_values:
    top_k = df_sorted.head(k)
    species_counts = top_k['species'].value_counts()
    predicted_species = species_counts.idxmax()
    
    print("\nTop", k, "neighbors:")
    print(top_k[['sepal length (cm)','sepal width (cm)','species','distance','rank']])
    print("Predicted species for k =", k, "→", predicted_species)

print("\nFinal conclusion: For input (", user_sepal_length, ",", user_sepal_width, ")")
print("Species predictions are shown above for all entered k values.")
```

```python
#Breast Cancer Dataset
from sklearn.datasets import load_breast_cancer
import pandas as pd, math

# Load dataset
cancer = load_breast_cancer()
df = pd.DataFrame(cancer.data, columns=cancer.feature_names)
df['target'] = [cancer.target_names[i] for i in cancer.target]

# Pick two features for simplicity
df = df[['mean radius','mean texture','target']]
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

print("Initial Table:")
print(df.head(10))

# User input
user_radius = float(input("Enter mean radius: "))
user_texture = float(input("Enter mean texture: "))

# Euclidean distance
distances = []
for i in range(len(df)):
    r = df.iloc[i,0]
    t = df.iloc[i,1]
    d = math.sqrt((user_radius - r)**2 + (user_texture - t)**2)
    distances.append(d)

df['distance'] = distances
df['rank'] = df['distance'].rank(method='first')
df_sorted = df.sort_values(by='distance').reset_index(drop=True)

print("\nTable with Distance and Rank:")
print(df_sorted.head(10))

# Multiple k values
k_values = input("Enter k values separated by commas (e.g. 1,3,5): ")
k_values = [int(k.strip()) for k in k_values.split(",")]

for k in k_values:
    top_k = df_sorted.head(k)
    species_counts = top_k['target'].value_counts()
    predicted = species_counts.idxmax()
    print("\nTop", k, "neighbors:")
    print(top_k)
    print("Predicted class for k =", k, "→", predicted)
```
